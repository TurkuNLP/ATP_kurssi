# Navigating the File System

Let's explore some basic Unix commands for navigating directories and managing files.

### 📁 Exploring directories

```bash
pwd                         # Show the current working directory
ls                          # List files and directories
ls -lh                      # List files with long information (-l) and human readable sizes (-h)
ls -lh sample_data          # List contents of a specific directory (note: relative path)
ls -lh /content/sample_data # List contents of a specific directory (note: absolute path)
```

### 📂 Creating and navigating directories

```bash
mkdir my_directory      # Create a new directory named "my_directory" under current working directory
cd my_directory         # Navigate into the directory (change location)
cd ..                   # Go to the parent directory (change location)
cd /content/sample_data # Go to the directory by using absolute path
```

**Note:** Typing `cd` without any arguments normally takes you to your home directory. In Google Colab, however, it takes you to /root, rather than the initial working directory (/content). If you get lost, you can use `cd /content` to return to the starting directory.

### 📄 Copying and moving files

```bash
cp sample_data/README.md .    # Copy a file into the current directory (. stands for here, the current working directory)
mv sample_data/README.md .    # Move a file into here
mv README.md my_readme.md     # Rename a file (move under one name to another)
```

The `mv` command can be used to move files between directories or rename files without moving. Note a difference between copy `cp` and move `mv`.


### 🗑️ Removing files and directories

```bash
rm my_readme.md         # Remove a file
rm -r my_directory      # Remove a directory (`-r`) and its contents
```

**Be careful:** `rm` permanently removes files and directories. The `-r` option removes directories recursively.


# Downloading and Exploring Text Files

Let's explore some basic Unix commands for downloading text files, viewing their contents, and counting lines, words, and characters.

### 🌐 Downloading files

```bash
wget https://example.com/file.txt    # Download a file from the web
```

This downloads the file and saves it as file.txt in your current working directory.


### 🔎 Viewing file contents

```bash
head -10 my_readme.md   # Show the first 10 lines of a file (prints to terminal)
less my_readme.md       # View the file interactively (opens a simple text editor)
```

**Note:** In `less`, press `Space` to scroll and `q` to exit.

### 🔢 Counting lines, words, and characters

```bash
wc myfile.txt       # Count lines, words, and bytes
wc -l myfile.txt    # Count lines
wc -w myfile.txt    # Count words
wc -m myfile.txt    # Count characters
```

**Note:** `wc` stands for *word count*. By default, it prints the number of lines, words, and bytes. The `-m` option counts characters instead of bytes, which can make a difference with UTF-8 text.
