## 3. First lines of Python

Python code is composed of lines, which are run sequentially. You can put as many lines to one cell as you like. Later lines are aware of variables created above.


`print()` shows a value below the cell. Here we have two kinds of values:

- **strings**: text, created by adding quotes, e.g. `"Alice"`
- **numbers**: integers like `3` and decimal numbers (*floats*) like `2.5`

By using hash (`#`), we can add comments to the code, any content after # **in the same line** does not affect what the code does

```python
print("Alice") # print a string "Alice"
print(3)       # print an integer 3
print(3 + 4)   # sum two integers (3 + 4), and print the sum (this line prints 7)
print(10 / 4)  # divide two numbers and print the outcome
print("Alice" + " in Wonderland") # sum two strings, and print the sum, summing string means concatenating them so (this line prints "Alice in Wonderland")
```

`type()` tells you what kind of value you have if you are unsure:

```python
print(type("Alice"))
print(type(3))
print(type(2.5))
```

A **variable** is a name for a value, so that you can store values for later use. The `=` sign gives the name to the value. After that, you can use the name instead of the value.

```python
sentence = "Alice in Wonderland"
number_a = 3
number_b = 4
print(sentence)
print(number_a + number_b)
```
